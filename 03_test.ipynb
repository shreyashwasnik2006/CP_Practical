```python
# student grade A-above-95 marks, B-80 to 95 marks, C-60 to 80 marks, D-40 to 60 marks, E-below 40 marks    
#if elif else format
marks = int(input('Enter your marks: '))
if marks >= 95:
    print('Your grade is A')
elif marks >= 80:
    print('Your grade is B')
elif marks >= 60:
    print('Your grade is C')
elif marks >= 40:
    print('Your grade is D')
else:
    print('Your grade is E')
```
